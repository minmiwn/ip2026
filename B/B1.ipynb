```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from utils.moduleAdjust import manual_cal_hist, normalization, cdf, equalization
```

```python
iname = "/home/chou/Project/ImageProcessing/study/self-study/image/img2.jpg"
img = cv2.imread(iname, 0)
plt.imshow(img, cmap='gray')
```

```python
# a
check1 = manual_cal_hist(img)
check2 = normalization(img)
check3 = cdf(img)
check4 = equalization(img)
check5 = normalization(check4)

plt.figure()
plt.bar(range(256), check2)
plt.xlabel("Gray Level")
plt.ylabel("Frequency")
plt.title("Normalization Histogram")
plt.show()

plt.figure()
plt.plot(range(256), check3)
plt.xlabel("Gray Level")
plt.ylabel("Frequency")
plt.title("CDF")
plt.show()
```

```python
# b 
h, w = img.shape
h_new, w_new = [int(x * 1.5) for x in img.shape[:2]]
print(f"{h, w} -> {h_new, w_new}")
```

```python
# c 
# original image
img_org = cv2.imread(iname)
plt.imshow(img_org[:,:,::-1])
plt.show()

# grayscale image
plt.imshow(img, cmap='gray')
plt.show()

# resized grayscale image
img_new = cv2.resize(img, (w_new, h_new))
plt.imshow(img_new, cmap='gray')

# histogram of the grayscale image
plt.figure()
plt.bar(range(256), check1)
plt.xlabel("Gray Level")
plt.ylabel("Frequency")
plt.title("Histogram")
plt.show()

# equalized grayscale image
# cv2.imwrite('equilization.png', check4.astype(np.uint8))
print(check4.shape)
print(check4.ndim)
plt.imshow(check4, cmap='gray')
plt.show()

# histogram of the equalized grayscale image
plt.figure()
plt.bar(range(256), check5)
plt.xlabel("Gray Level")
plt.ylabel("Frequency")
plt.title("Equilization Histogram")
plt.show()

# grayscale image and equalized grayscale image side by side for comparison
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(img_org[:,:,::-1], cmap='gray')
plt.title("Original Grayscale")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(check4, cmap='gray')
plt.title("Equalized Grayscale")
plt.axis('off')

# plt.tight_layout()
plt.show()
```
