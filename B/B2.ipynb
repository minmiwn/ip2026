```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from moduleAdjust import manual_cal_hist, normalization, cdf, equalization, bw, transform
```

```python
iname = '/home/chou/Project/ImageProcessing/study/self-study/image/kat.png'
img = cv2.imread(iname)
plt.imshow(img[:,:,::-1])
```

```python
# g(i,j) = αf(i,j) + β
# α - gain - contrast (độ tương phản) 
# β - bias - brightness (độ sáng)
# input overexposure_img -> function -> output
# dtype vs astype() 
# EXPLAIN:
# +/- b - tịnh tiến lên/xuống toàn bộ grayscale - increase/decrease bright - Overexposure/Underexposure 
# >/< 1 - thay đổi khoảng cách (độ dốc) giữa các mức xám - kéo dãn ra/co lại - High Contrast/Low Contrast
# vùng sáng sáng hơn/vùng tối giảm ít hoặc giữ nguyên - gắt hơn, vùng sáng tối phân rõ ràng 
# có lớp sương mù phủ 

res1 = transform(img, 1.0, 50)
res2 = transform(img, 1.0, -50)
res3 = transform(img, 1.5, 0)
res4 = transform(img, 0.5, 0)

plt.figure(figsize=(10, 5))

plt.subplot(1, 5, 1)
plt.imshow(img[:,:,::-1])
plt.title("Original")
plt.axis('off')

plt.subplot(1, 5, 2)
plt.imshow(res1[:,:,::-1])
plt.title("[1.0, 50]")
plt.axis('off')

plt.subplot(1, 5, 3)
plt.imshow(res2[:,:,::-1])
plt.title("[1.0, -50]")
plt.axis('off')

plt.subplot(1, 5, 4)
plt.imshow(res3[:,:,::-1])
plt.title("[1.5, 0]")
plt.axis('off')

plt.subplot(1, 5, 5)
plt.imshow(res4[:,:,::-1])
plt.title("[0.5, 0]")
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
img_gray0 = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
img_gray1 = cv2.cvtColor(res1, cv2.COLOR_BGR2GRAY)
img_gray2 = cv2.cvtColor(res2, cv2.COLOR_BGR2GRAY)
img_gray3 = cv2.cvtColor(res3, cv2.COLOR_BGR2GRAY)
img_gray4 = cv2.cvtColor(res4, cv2.COLOR_BGR2GRAY)

hist0 = manual_cal_hist(img_gray0)
hist1 = manual_cal_hist(img_gray1)
hist2 = manual_cal_hist(img_gray2)
hist3 = manual_cal_hist(img_gray3)
hist4 = manual_cal_hist(img_gray4)

plt.figure()
plt.bar(range(256), hist0)
plt.xlabel("Gray Level")
plt.ylabel("Frequency")
plt.title("Histogram")
plt.show()

plt.figure()
plt.bar(range(256), hist1)
plt.xlabel("Gray Level")
plt.ylabel("Frequency")
plt.title("Histogram")
plt.show()

plt.figure()
plt.bar(range(256), hist2)
plt.xlabel("Gray Level")
plt.ylabel("Frequency")
plt.title("Histogram")
plt.show()

plt.figure()
plt.bar(range(256), hist3)
plt.xlabel("Gray Level")
plt.ylabel("Frequency")
plt.title("Histogram")
plt.show()

plt.figure()
plt.bar(range(256), hist4)
plt.xlabel("Gray Level")
plt.ylabel("Frequency")
plt.title("Histogram")
plt.show()
```
